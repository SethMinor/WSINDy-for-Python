```python
from dysts.flows import Lorenz
import matplotlib.pyplot as plt

model = Lorenz()
sol = model.make_trajectory(500)

print(type(sol))
print(sol.shape)

plt.plot(sol[:,0])
plt.plot(sol[:,1])
plt.plot(sol[:,2])
plt.show()
```

```python
import inspect

print(inspect.getsource(model._rhs))
print(model.params)
```
